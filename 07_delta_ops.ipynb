```python
dbutils.widgets.text("catalog", "vp_rm20040829")
catalog = dbutils.widgets.get("catalog")
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {catalog}.ops.time_travel_check AS
SELECT
    curr.charger_id,
    prev.max_kw          AS max_kw_before_batch2,
    curr.max_kw          AS max_kw_current,
    prev.is_active       AS is_active_before_batch_2,
    curr.is_active       AS is_active_current
FROM {catalog}.silver.chargers                        AS curr
JOIN {catalog}.silver.chargers VERSION AS OF {version_before_batch2} AS prev
    ON curr.charger_id = prev.charger_id
WHERE curr.charger_id IN ('CH004', 'CH011', 'CH027')
""")

display(spark.table(f"{catalog}.ops.time_travel_check"))
```
